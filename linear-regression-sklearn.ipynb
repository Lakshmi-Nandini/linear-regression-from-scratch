```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
X_train = np.array([
    [800, 2],
    [1000, 2],
    [1200, 3],
    [1500, 3],
    [1800, 4]
], dtype=float)

y_train = np.array([40, 50, 65, 80, 100], dtype=float)

X_test = np.array([
    [900, 2],
    [1300, 3],
    [1700, 4]
], dtype=float)

y_test = np.array([45, 70, 95], dtype=float)
```

```python
model = LinearRegression()

model.fit(X_train, y_train)
```

```python
print("Weight for area:", model.coef_[0])
print("Weight for bedrooms:", model.coef_[1])
print("Intercept:", model.intercept_)
```

```python
train_predictions = model.predict(X_train)

print("Actual prices:   ", y_train)
print("Predicted prices:", np.round(train_predictions, 2))

train_mse = mean_squared_error(y_train, train_predictions)

print("Training MSE:", train_mse)
```

```python
test_predictions = model.predict(X_test)

print("Actual prices:   ", y_test)
print("Predicted prices:", np.round(test_predictions, 2))

test_mse = mean_squared_error(y_test, test_predictions)

print("Test MSE:", test_mse)
```
